neuraon is the basic unit of the neural network or we can say that is the smallest neural neetwork

```
import torch
import torch.nn as nn #iske andar basically hamare building blocks hote hai jaise ki hame koi activation function cuse karna hai layers create karni hai
# // this is tha alias that we have


```

```
# creation of input
torch.manual_seed(42) # controlling the randomness
inputs = torch.tensor([1.0,2.0 , 3.0]) #input hame float main dena hota hai

#  tesor creates multidimentional array


```

```
#creating a neuron

neuron = nn.Linear(in_features = 3 , out_features = 1 ) # defining the number of input and numbere of output for the neuron
```

```
#storing the output of the neuron

output = neuron(inputs);
#  printing the output
output
```

```
#bias
neuron.bias
```

weighted sum formula = $$ ∑_{i=1} ^{n} x_i. w_i + bias$$


```
#varifying the output using weighted sum formula
#we use @ for dot product in python
neuron.weight @inputs +neuron.bias
```
